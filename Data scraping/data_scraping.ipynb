```python
import re
from datetime import date
from io import StringIO

import pandas as pd
import requests
from bs4 import BeautifulSoup

url = "https://www.ccny.cuny.edu/registrar/fall"

response = requests.get(url)
response.raise_for_status()

soup = BeautifulSoup(response.text, "html.parser")
table = soup.find("table")

calendar_df = pd.read_html(StringIO(str(table)))[0]
calendar_df.columns = ["date_text", "dow", "text"]

def to_python_date(value):
    value = str(value).replace("\xa0", " ").strip()
    match = re.search(r"([A-Za-z]+)\s+(\d{1,2})(?:,\s*(\d{4}))?", value)

    month, day, year = match.groups()
    year = int(year) if year else (2022 if month == "January" else 2021)

    return date(year, int(day), pd.to_datetime(month, format="%B").month)

calendar_df.index = pd.Index(
    calendar_df["date_text"].map(to_python_date),
    name="date"
)

calendar_df = calendar_df[["dow", "text"]]
calendar_df
```
